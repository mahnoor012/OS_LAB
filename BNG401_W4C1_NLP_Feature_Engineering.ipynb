# 1. NLP Pipeline Perspective

A complete NLP system follows:

```
Raw Text
    ↓
Text Preprocessing
    ↓
Feature Engineering
    ↓
Numerical Representation
    ↓
Machine Learning Model
    ↓
Prediction
```

In this class we focus on:

```
Processed Text → Useful Features
```

The goal is to understand what information can be extracted from text before converting it into vectors.


````python
# Example Bangla text dataset

documents = [
    "আমি বাংলা শিখি",
    "আমি বাংলা পড়ি",
    "বাংলা ভাষা সুন্দর"
]

documents
````

# 2. Token Analysis

Tokenization was introduced earlier.

Now we use tokens for analysis.

Example:

Sentence:

```
আমি বাংলা শিখি
```

Tokens:

```
আমি
বাংলা
শিখি
```

Feature engineering starts by analyzing these linguistic units.


````python
# Tokenizing documents for analysis

tokenized_documents = [
    sentence.split()
    for sentence in documents
]

tokenized_documents
````

# 3. Vocabulary Analysis

A vocabulary is the collection of unique words appearing in a dataset.

Why is vocabulary important?

Because NLP systems need to understand:

- what words exist in the dataset
- how diverse the language is
- which words appear frequently

Example:

Dataset:

```
আমি বাংলা শিখি
আমি বাংলা পড়ি
```

Vocabulary:

```
আমি
বাংলা
শিখি
পড়ি
```


````python
# Creating vocabulary

vocabulary = set()

for document in tokenized_documents:
    for word in document:
        vocabulary.add(word)

vocabulary
````

# 4. Vocabulary Size

Vocabulary size is an important property of NLP datasets.

A small vocabulary:

- easier processing
- fewer features

A large vocabulary:

- more language information
- higher computational complexity

Real datasets may contain thousands or millions of unique tokens.


````python
# Vocabulary size

vocabulary_size = len(vocabulary)

print("Vocabulary size:", vocabulary_size)
````

# 5. Word Frequency Analysis

Word frequency is one of the simplest text features.

Why?

Frequent words may indicate:

- important topics
- common patterns
- possible stopwords

Example:

In news articles:

```
বাংলাদেশ
সরকার
নির্বাচন
```

may appear frequently.


````python
from collections import Counter

all_words = []

for document in tokenized_documents:
    all_words.extend(document)

word_frequency = Counter(all_words)

word_frequency
````

# 6. Finding Frequent Words

Frequency analysis helps us understand the dataset.

Questions:

- Which words dominate the corpus?
- Are there repeated common words?
- Are there unusual words?

This analysis is useful before building NLP models.


````python
# Most frequent words

word_frequency.most_common(5)
````

# 7. Document Length Analysis

Document length can be an important feature.

Examples:

Short text:
- social media posts
- search queries

Long text:
- articles
- reviews

Understanding document length helps select appropriate NLP methods.


````python
# Calculate document lengths

for i, document in enumerate(tokenized_documents):
    print(
        "Document",
        i+1,
        "length:",
        len(document),
        "tokens"
    )
````

# 8. Unique Word Ratio

A document with many repeated words may contain less information.

Unique word ratio:

\[
Unique\ Words / Total\ Words
\]

Higher ratio:
- more vocabulary diversity

Lower ratio:
- repeated patterns


````python
for document in tokenized_documents:
    total_words = len(document)
    unique_words = len(set(document))

    ratio = unique_words / total_words

    print("Unique word ratio:", ratio)
````

# 10. Practical Dataset Exploration

Before training an NLP model, we usually inspect:

✓ Vocabulary size  
✓ Frequent words  
✓ Document length  
✓ Word distribution  
✓ Data quality

This process helps us understand the dataset before feature representation.


````python
# Simple dataset summary function

def dataset_summary(documents):

    total_documents = len(documents)

    all_tokens = [
        word
        for doc in documents
        for word in doc.split()
    ]

    return {
        "documents": total_documents,
        "total_tokens": len(all_tokens),
        "unique_words": len(set(all_tokens))
    }


dataset_summary(documents)
````
