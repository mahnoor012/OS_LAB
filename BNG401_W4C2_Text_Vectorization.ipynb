# 1. Example Bangla Corpus

A machine learning model requires numerical input.

We start with text documents and transform them into feature representations.


```
documents = [
    "আমি বাংলা শিখি",
    "আমি বাংলা পড়ি",
    "বাংলা ভাষা সুন্দর"
]

documents
```

# 2. Count Vectorization

Bag of Words represents text using word occurrence.

The algorithm:

1. Create vocabulary
2. Assign index
3. Count word occurrences
4. Create feature matrix


```
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()

bow_matrix = count_vectorizer.fit_transform(documents)

print(count_vectorizer.get_feature_names_out())
print(bow_matrix.toarray())
```

# 3. Understanding the Feature Matrix

Rows:
Documents

Columns:
Vocabulary words

Values:
Number of occurrences

This matrix becomes input for classical ML models.


```
import pandas as pd

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)

bow_df
```

# 4. TF-IDF Representation

TF-IDF improves upon Count Vectorization.

It gives higher importance to words that are:

- frequent in a document
- rare across documents

Formula:

TF-IDF = TF × IDF


```
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

print(tfidf_vectorizer.get_feature_names_out())
print(tfidf_matrix.toarray())
```

# 5. Comparing BoW and TF-IDF

Count Vector:

- raw frequency
- simple interpretation

TF-IDF:

- weighted importance
- reduces common word influence


```
comparison = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

comparison
```

# 6. N-gram Generation

N-grams preserve local word order.

Example:

Bigram:

আমি বাংলা

Trigram:

আমি বাংলা শিখি


```
from sklearn.feature_extraction.text import CountVectorizer

bigram_vectorizer = CountVectorizer(
    ngram_range=(1,2)
)

bigram_matrix = bigram_vectorizer.fit_transform(documents)

print(bigram_vectorizer.get_feature_names_out())
print(bigram_matrix.toarray())
```

# 7. Observations

Bag of Words:
- easy
- interpretable
- ignores context

TF-IDF:
- improves weighting
- still lacks semantics

N-grams:
- capture local order
- increase vocabulary size

These limitations motivate word embeddings and neural NLP.
